```python
from pathlib import Path
import os
import time
from datetime import date
from io import StringIO

import pandas as pd
import requests


%load_ext autoreload
%autoreload 2

from USGS_Functions import (
    get_usgs_continuous,
    get_start_date_instantaneous
)
# ============================================================
# PATHS
cwd = os.getcwd() ; script_dir = Path.cwd() 
```

```python
USGS_API_KEY = "cdEidIqo5qAwbyENW7VinWreN8JbBlsvjcbrPtsm"

output_folder = script_dir / "Continuous_stream_data"
output_folder.mkdir(parents=True, exist_ok=True)


output_folder = script_dir / "Continuous_reservoir_data"
output_folder.mkdir(parents=True, exist_ok=True)


USGS_CONTINUOUS_URL = (
    "https://api.waterdata.usgs.gov/"
    "ogcapi/v1/collections/continuous/items"
)
```

# Select Gauges

```python
excel_file = script_dir / "active_stream_gauges_NWS_stages.xlsx"

# ============================================================
# READ GAUGE TABLE
# ============================================================

gauges = pd.read_excel(
    excel_file,
    dtype={"SITENO": str}
)

gauges["SITENO"] = (
    gauges["SITENO"]
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)

print(gauges[["SITENO", "SITENAME"]].head())
print(f"\nNumber of gauges: {len(gauges)}")
```

# Get the start year of instantaneous gauge, download data into csv of discharge and stage

```python
# ============================================================
# GET CONTINUOUS / INSTANTANEOUS START DATE FOR EACH GAUGE
# ============================================================

gauges["Continuous_Start"] = None

for idx, row in gauges.iterrows():

    site_no = str(row["SITENO"]).strip()
    try:

        start_date = get_start_date_instantaneous(
            usgs_id=site_no,
            api_key=USGS_API_KEY,
            parameter_codes=(
                "00060",   # discharge
                "00065"    # gage height
            )
        )

        gauges.loc[
            idx,
            "Continuous_Start"
        ] = start_date

    except Exception as e:

        print(
            f"FAILED USGS {site_no}: {e}"
        )

        gauges.loc[
            idx,
            "Continuous_Start"
        ] = None


# ============================================================
# SHOW RESULTS
# ============================================================

display(
    gauges[
        [
            "SITENO",
            "SITENAME",
            "Continuous_Start"
        ]
    ]
)
```

# Pull all continuos data

```python
# ============================================================
# PULL CONTINUOUS USGS DATA
# ============================================================

for site_no in gauges["SITENO"].astype(str):

    # --------------------------------------------------------
    # Get continuous record start
    # --------------------------------------------------------

    start = get_start_date_instantaneous(
        site_no,
        api_key=USGS_API_KEY
    )

    if start is None:
        continue


    # --------------------------------------------------------
    # Pull discharge + stage
    # --------------------------------------------------------

    data = get_usgs_continuous(
        site_no=site_no,
        api_key=USGS_API_KEY,
        output_folder=output_folder,
        start_year=start.year,
        end_year=2026
    )
```
