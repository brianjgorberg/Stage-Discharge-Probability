```python
import pandas as pd
from pathlib import Path

from pathlib import Path
#===================================
# FUNCTION: SAVE GAUGE METADATA
# ============================================================

def save_gauge_metadata(gauge_ids, input_file, output_file):

    # Read metadata (keep gauge IDs as strings)
    df = pd.read_csv(
        input_file,
        dtype={"STA_ID": str}
    )

    # Convert requested IDs to strings
    gauge_ids = [str(gauge).strip() for gauge in gauge_ids]

    # Select matching gauges
    selected = df[
        df["STA_ID"].isin(gauge_ids)
    ].copy()

    # Preserve requested gauge order
    selected["STA_ID"] = pd.Categorical(
        selected["STA_ID"],
        categories=gauge_ids,
        ordered=True
    )

    selected = selected.sort_values("STA_ID")

    # Convert back to strings
    selected["STA_ID"] = selected["STA_ID"].astype(str)

    # Create output directory
    output_file = Path(output_file)
    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    # Save selected metadata
    selected.to_csv(output_file, index=False)

    # Identify missing gauge IDs
    missing = set(gauge_ids) - set(selected["STA_ID"])

    print(f"Saved {len(selected)} gauges")
    print(f"Output: {output_file}")

    if missing:
        print(f"Gauge IDs not found: {sorted(missing)}")

    return selected


# ============================================================
```

```python
# Current notebook directory
script_dir = Path.cwd()

# Input CSV
input_file = script_dir / "Hawaii_StreamStats_All_Gages" / "Hawaii_StreamStats_Gages_Peak_Record.csv"

# Output folder
output_dir = script_dir / "Hawaii_StreamStats_All_Gages"

# Output CSV
output_file = script_dir / "SAVE_DATA" / "USGS_Gauge_Metadata.csv"


gauge_ids = [
    "16103000",  # Hanalei
    "16294100",  # Waiahole
    "16210500",  # Waialua
]
gauge_metadata = save_gauge_metadata(
    gauge_ids,
    input_file,
    output_file
)

display(gauge_metadata)
# =========================
```
