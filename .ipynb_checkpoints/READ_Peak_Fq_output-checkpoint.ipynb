```python
from pathlib import Path
import os
import time
from datetime import date
from io import StringIO

import pandas as pd
import requests
from USGS_Functions import (
    get_usgs_continuous,
    get_start_date_instantaneous,
    read_nwis_peak
)
# ============================================================
# PATHS
cwd = os.getcwd() ; script_dir = Path.cwd() 
# Output folder
output_dir = script_dir / "SAVE_DATA"
output_folder =   (  output_dir  / "Stream_Gauge_Annual_Peaks" )
output_folder.mkdir(
    parents=True,
    exist_ok=True
)
```

# Read Peak Fq paths

```python
# Peak_FQ folder
peak_fq_dir = script_dir / "Peak_Fq"

# Get all folder names inside Peak_FQ
folder_names = sorted([
    folder.name
    for folder in peak_fq_dir.iterdir()
    if folder.is_dir()
])

print(f"Number of gauges: {len(folder_names)}")
```

# Organize probabilities in a metadata sheet

```python
# ============================================================
# READ QNT FILES
# ============================================================

probabilities = {}
peaks = {}
for site_no in folder_names:

    # --------------------------------------------------------
    # Gauge folder
    # --------------------------------------------------------

    gauge_dir = (
        peak_fq_dir
        / site_no
    )


    # --------------------------------------------------------
    # QNT file
    # --------------------------------------------------------

    qnt_file = (
        gauge_dir
        / f"{site_no}_qnt.csv"
    )


    # --------------------------------------------------------
    # Read
    # --------------------------------------------------------

    qnt = pd.read_csv(
        qnt_file
    )

    peak_data = read_nwis_peak(gauge_dir / f"{site_no}_nwis_peak.txt")

    peaks[site_no] = peak_data
    probabilities[site_no] = qnt

```

```python
# ============================================================
# CREATE META PROBABILITY DATAFRAME
# ============================================================

meta_probability = pd.DataFrame()

for site_no, df in probabilities.items():

    # --------------------------------------------------------
    # Get probability + estimate
    # --------------------------------------------------------

    gauge_probability = (
        df[
            [
                "EXC_Prob",
                "Estimate"
            ]
        ]
        .set_index("EXC_Prob")
    )


    # --------------------------------------------------------
    # Rename estimate to site number
    # --------------------------------------------------------

    gauge_probability = gauge_probability.rename(
        columns={
            "Estimate": str(site_no)
        }
    )


    # --------------------------------------------------------
    # Merge
    # --------------------------------------------------------

    if meta_probability.empty:

        meta_probability = gauge_probability.copy()

    else:

        meta_probability = meta_probability.join(
            gauge_probability,
            how="outer"
        )


# ============================================================
# SORT BY EXCEEDANCE PROBABILITY
# ============================================================

meta_probability = meta_probability.sort_index(
    ascending=False
)


# ============================================================
# SAVE
# ============================================================
# Output folder

output_file = (
    output_dir
    / "Peak_Fq_USGS_probability.csv"
)

meta_probability.to_csv(
    output_file
)


meta_probability
```

```python
import pandas as pd
from pathlib import Path

# ============================================================
# SETTINGS
# ============================================================

watershed = "Waialua_weighted"

output_dir = Path("SAVE_DATA")
output_dir.mkdir(parents=True, exist_ok=True)

# ============================================================
# FUNCTION: SAVE PROBABILITY CURVE
# ============================================================

def save_probability_function(
    watershed,
    curve_discharge,
    curve_probability,
    output_dir="SAVE_DATA"
):

    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    output_file = (
        output_dir / f"{watershed}_probability_function.csv"
    )

    df = pd.DataFrame({
        "discharge_cfs": curve_discharge,
        "EXC_Prob": curve_probability,
        "return_period_years": 1 / curve_probability
    })

    df.to_csv(output_file, index=False)

    print(f"Saved: {output_file}")

    return df


# ============================================================
# RUN
# ============================================================

saved_probability = save_probability_function(
    watershed,
    curve_discharge,
    curve_probability,
    output_dir
)

display(saved_probability)
```

# Pull annual peaks and save as csv for each gauge

```python
# ============================================================
# SAVE ANNUAL PEAK DATA
# ============================================================

output_folder = (
    output_dir
    / "Stream_Gauge_Annual_Peaks"
)

output_folder.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# LOOP THROUGH PEAK DICTIONARY
# ============================================================

for site_no, peak_data in peaks.items():

    # --------------------------------------------------------
    # Output file
    # --------------------------------------------------------

    output_file = (
        output_folder
        / f"{site_no}_peaks.csv"
    )


    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    peak_data.to_csv(
        output_file,
        index=False
    )


print(
    f"Saved {len(peaks)} gauges"
)

# ============================================================
# CREATE ANNUAL PEAK DATAFRAME
# ============================================================

```
